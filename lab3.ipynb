# Road Network Analysis using DFS
This program implements both functions:
1. Checks if a path exists between two cities.
2. Identifies all separate connected groups of cities (connected components).

```python
# Define the road network as an adjacency list
road_network = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A'],
    'D': ['B'],
    'E': ['F'],
    'F': ['E'],
    'G': []  # Isolated city
}

def dfs_path_exists(graph, start, end, visited=None):
    if visited is None:
        visited = set()
    if start == end:
        return True
    visited.add(start)
    for neighbor in graph.get(start, []):
        if neighbor not in visited:
            if dfs_path_exists(graph, neighbor, end, visited):
                return True
    return False

def find_connected_components(graph):
    visited = set()
    components = []

    def dfs_collect(node, component):
        visited.add(node)
        component.append(node)
        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                dfs_collect(neighbor, component)

    for node in graph:
        if node not in visited:
            component = []
            dfs_collect(node, component)
            components.append(component)
    return components

# --- Demonstration ---
print("--- Task 1: Check Route Existence ---")
print("Route exists between A and D:", dfs_path_exists(road_network, 'A', 'D'))
print("Route exists between A and E:", dfs_path_exists(road_network, 'A', 'E'))

print("\n--- Task 2: Find Connected Components ---")
components = find_connected_components(road_network)
for idx, component in enumerate(components, 1):
    print(f"Group {idx}: {component}")
```
