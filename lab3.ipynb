```python
# Python program using DFS for road network connectivity in West Bengal

class RoadNetwork:
    def __init__(self):
        # Adjacency list representation of the graph
        self.graph = {}

    def add_city(self, city):
        if city not in self.graph:
            self.graph[city] = []

    def add_road(self, city1, city2):
        self.add_city(city1)
        self.add_city(city2)
        # Since roads are two-way
        self.graph[city1].append(city2)
        self.graph[city2].append(city1)

    # 1. Check whether a route exists between two given cities using DFS
    def route_exists(self, start, destination):
        if start not in self.graph or destination not in self.graph:
            return False

        visited = set()
        return self._dfs_search(start, destination, visited)

    def _dfs_search(self, current, destination, visited):
        if current == destination:
            return True

        visited.add(current)
        for neighbor in self.graph[current]:
            if neighbor not in visited:
                if self._dfs_search(neighbor, destination, visited):
                    return True
        return False

    # 2. Find all separate groups of cities (Connected Components)
    def find_connected_components(self):
        visited = set()
        components = []

        for city in self.graph:
            if city not in visited:
                component = []
                self._dfs_collect(city, visited, component)
                components.append(component)
        return components

    def _dfs_collect(self, current, visited, component):
        visited.add(current)
        component.append(current)
        for neighbor in self.graph[current]:
            if neighbor not in visited:
                self._dfs_collect(neighbor, visited, component)

# Initialize the West Bengal road network
network = RoadNetwork()

# Group 1: South-Central Bengal Cluster
network.add_road("Kolkata", "Howrah")
network.add_road("Kolkata", "Kharagpur")
network.add_road("Kharagpur", "Medinipur")
network.add_road("Howrah", "Bardhaman")

# Group 2: North Bengal Cluster
network.add_road("Siliguri", "Darjeeling")
network.add_road("Siliguri", "Kalimpong")
network.add_road("Jalpaiguri", "Cooch Behar")
network.add_road("Siliguri", "Jalpaiguri")

# Group 3: Isolated/Other Districts Cluster
network.add_road("Asansol", "Durgapur")
network.add_road("Purulia", "Bankura")

# Group 4: Completely isolated city
network.add_city("Malda")

# --- Test Cases ---
print("=== ROAD NETWORK CONNECTIVITY TEST ===\n")

# Test 1: Route existence check
city_a, city_b = "Kolkata", "Bardhaman"
print(f"Route exists between '{city_a}' and '{city_b}': {network.route_exists(city_a, city_b)}")

city_c, city_d = "Kolkata", "Siliguri"
print(f"Route exists between '{city_c}' and '{city_d}': {network.route_exists(city_c, city_d)}")

city_e, city_f = "Asansol", "Durgapur"
print(f"Route exists between '{city_e}' and '{city_f}': {network.route_exists(city_e, city_f)}")

print("\n=== CONNECTED COMPONENTS OF CITIES ===")
components = network.find_connected_components()
for idx, comp in enumerate(components, 1):
    print(f"Group {idx}: {', '.join(comp)}")
```
