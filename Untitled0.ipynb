```python
import collections
import networkx as nx
import matplotlib.pyplot as plt

# 1. Store the graph using an adjacency list
# Representing the tree with A as root and its children
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

# 2. Perform a BFS traversal
def bfs(graph, start_node):
    visited = set()
    queue = collections.deque([start_node])
    traversal_order = []
    node_visit_order = {}
    order_counter = 1

    while queue:
        node = queue.popleft()
        if node not in visited:
            visited.add(node)
            traversal_order.append(node)
            node_visit_order[node] = order_counter
            order_counter += 1
            for neighbor in graph[node]:
                if neighbor not in visited:
                    queue.append(neighbor)
    return traversal_order, node_visit_order

# 3. Perform a DFS traversal
def dfs(graph, start_node):
    visited = set()
    stack = [start_node]
    traversal_order = []
    node_visit_order = {}
    order_counter = 1

    while stack:
        node = stack.pop()
        if node not in visited:
            visited.add(node)
            traversal_order.append(node)
            node_visit_order[node] = order_counter
            order_counter += 1
            # Add neighbors in reverse order to simulate left-to-right traversal
            # when popping from the stack.
            for neighbor in reversed(graph[node]):
                if neighbor not in visited:
                    stack.append(neighbor)
    return traversal_order, node_visit_order

# Perform traversals starting from node 'A'
bfs_order, bfs_node_order = bfs(graph, 'A')
dfs_order, dfs_node_order = dfs(graph, 'A')

# 4. Print both traversal orders
print("BFS Traversal Order:", bfs_order)
print("DFS Traversal Order:", dfs_order)

# 5. Draw two separate tree diagrams
def draw_graph_with_order(graph_dict, node_order, title):
    G = nx.DiGraph() # Use DiGraph for directed edges
    for node, neighbors in graph_dict.items():
        for neighbor in neighbors:
            G.add_edge(node, neighbor)

    # Use spring_layout for a general, readable layout. Seed for reproducibility.
    pos = nx.spring_layout(G, seed=42)

    # Create labels with visit order
    labels = {node: f"{node} ({order})" for node, order in node_order.items()}

    plt.figure(figsize=(8, 6))
    nx.draw_networkx_nodes(G, pos, node_color='lightblue', node_size=2000)
    nx.draw_networkx_edges(G, pos, edge_color='gray', arrows=True, arrowsize=20)
    nx.draw_networkx_labels(G, pos, labels, font_size=10, font_weight='bold')
    plt.title(title)
    plt.axis('off') # Hide axes
    plt.show()

# Drawing BFS traversal visualization
draw_graph_with_order(graph, bfs_node_order, "BFS Traversal Order")

# Drawing DFS traversal visualization
draw_graph_with_order(graph, dfs_node_order, "DFS Traversal Order")
```
